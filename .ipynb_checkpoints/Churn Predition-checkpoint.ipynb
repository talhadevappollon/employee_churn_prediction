```python
# Step 1: Import required libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, confusion_matrix
```

```python
df= pd.read_csv("Dept_Pred_HR_Attrition.csv")
df.head()
```

```python
df.info()
```

```python
# Step 3.1: Create churn target variable
df['churn'] = df['Attrition'].map({'Yes': 1, 'No': 0})

# Verify conversion
df['churn'].value_counts()
```

```python
# Step 3.2: Drop unnecessary columns
drop_cols = [
    'Attrition',          # replaced by churn
    'Attrition Date',     # future information (leakage)
    'EmployeeNumber',     # unique identifier
    'EmployeeCount',      # constant
    'StandardHours',      # constant
    'Over18',             # constant
    'Random Number'       # noise
]

df = df.drop(columns=drop_cols)
```

```python
df.head()
```

```python
df.info()
```

```python
# Step 4.1: Identify categorical columns
categorical_cols = df.select_dtypes(include='object').columns
categorical_cols
```

```python
# Step 4.2: One-hot encode categorical variables
df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)
```

```python
# Check new shape and sample
df_encoded.shape
df_encoded.head()
```

```python
# Step 5.1: Separate features (X) and target (y)
X = df_encoded.drop('churn', axis=1)
y = df_encoded['churn']
# Step 5.2: Check churn balance
y.value_counts(normalize=True)
```

```python
# Step 5.3: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
# Step 6.1: Train Decision Tree model
dt_model = DecisionTreeClassifier(
    max_depth=4,          # keeps model interpretable
    min_samples_leaf=30,  # avoids noisy splits
    random_state=42
)

dt_model.fit(X_train, y_train)
```

```python
# Step 6.2: Predict on test data
y_pred = dt_model.predict(X_test)
# Step 6.3: Evaluation metrics
print(classification_report(y_test, y_pred))
```

```python
# Confusion Matrix
confusion_matrix(y_test, y_pred)
```

```python
# Step 7: Visualize decision tree

plt.figure(figsize=(24, 12))
plot_tree(
    dt_model,
    feature_names=X.columns,
    class_names=['Stayed', 'Left'],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.show()
```

```python
# Step 8.1: Feature importance
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': dt_model.feature_importances_
}).sort_values(by='Importance', ascending=False)

feature_importance.head(10)
```

```python
# Step 8.2: Plot top 10 features
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Importance',
    y='Feature',
    data=feature_importance.head(10)
)
plt.title("Top 10 Churn Drivers")
plt.show()

# “These variables have the highest impact on employee attrition.
# The model consistently splits on these factors first, meaning they are the strongest churn signals.”

# Typical top drivers you’ll likely see:

# OverTime

# JobSatisfaction

# MonthlyIncome

# YearsAtCompany

# WorkLifeBalance

# JobLevel
```
